**Aim**

To implement page frame with FIFO and LRU system and calculate the number of hits and misses.

**Problem Statement**

Consider a system with 3 page frames and the following reference string containing 20 page references.
Calculate the total page faults and page hits using FIFO and LRU page replacement policies.

```python
# ref_str = '7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2, 1, 2, 0, 1, 7, 0, 1'
# ref_arr = ref_str.split(',')
# ref_arr = [int(x) for x in ref_arr]
ref_arr = [7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2, 1, 2, 0, 1, 7, 0, 1]
frame_no = 3
```

1. FIFO Page Replacement

```python
page = [-1, -1, -1]
page_ptr = miss = hit = 0
for i in range(len(ref_arr)):
    if page_ptr == 3: # Loop around condition
        page_ptr = 0
    if ref_arr[i] in page: # if exists in page then a hit
        print(i, 'H', end=" ")
        hit+=1
    else:
        page[page_ptr] = ref_arr[i]
        print(i, 'F', end=" ")
        miss+=1
        page_ptr+=1
    print(page)

print("Total no of Miss/Faults = ", miss)
print("Total no of Hit/Non Faults = ", hit)
print("Miss/Hit Ratio: ", miss/hit)
```

2. LRU Page Replacement.

```python
page = [-1, -1, -1]
page_use = [-1, -1, -1]
miss = 0
hit = 0

for i in range(len(ref_arr)):
    if ref_arr[i] in page: # if exists in page then a hit
        page_use[page.index(ref_arr[i])] = i
        print(i, 'H', end=" ")
        hit+=1
    else:
        replace = page_use.index(min(page_use))
        page[replace] = ref_arr[i]
        page_use[replace] = i
        print(i, 'F', end=" ")
        miss+=1
    print(page, page_use)

print("Total no of Miss/Faults = ", miss)
print("Total no of Hit/Non Faults = ", hit)
print("Miss/Hit Ratio: ", miss/hit)
```
