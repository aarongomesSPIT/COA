```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the final simulation output
cores = np.array([1, 2, 4, 8])
cycles = np.array([822085, 822085, 822085, 822085])
t1 = cycles[0]
speedup = t1 / cycles
ideal_speedup = cores

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# ---------------------------------------------------------
# Image 1: Cycles vs. Number of Cores
# ---------------------------------------------------------
fig1, ax1 = plt.subplots(figsize=(8, 5), dpi=120)
ax1.plot(cores, cycles, marker='o', color='#1f77b4', linewidth=2.5, markersize=8, label='Measured Cycles')
ax1.set_title('Cycles vs. Number of Cores', fontsize=13, fontweight='bold', pad=12)
ax1.set_xlabel('Number of Cores (n)', fontsize=11)
ax1.set_ylabel('Total Cycles', fontsize=11)
ax1.set_xticks(cores)
ax1.set_ylim(0, 1000000)
ax1.yaxis.set_major_formatter('{x:,.0f}')

for x, y in zip(cores, cycles):
    ax1.annotate(f'{y:,}', (x, y), textcoords="offset points", xytext=(0, 10), ha='center', fontsize=9, fontweight='bold', color='#1f77b4')

ax1.legend(loc='upper right', frameon=True)
ax1.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('cycles_vs_cores.png', dpi=300)
plt.show()

# ---------------------------------------------------------
# Image 2: Speedup vs. Number of Cores
# ---------------------------------------------------------
fig2, ax2 = plt.subplots(figsize=(8, 5), dpi=120)
ax2.plot(cores, speedup, marker='s', color='#d62728', linewidth=2.5, markersize=8, label='Measured Speedup (S)')
ax2.plot(cores, ideal_speedup, linestyle='--', color='#2ca02c', linewidth=2, label='Ideal Speedup (S = n)')
ax2.set_title('Speedup vs. Number of Cores', fontsize=13, fontweight='bold', pad=12)
ax2.set_xlabel('Number of Cores (n)', fontsize=11)
ax2.set_ylabel('Speedup (T1 / Tn)', fontsize=11)
ax2.set_xticks(cores)
ax2.set_ylim(0, 9)
ax2.set_yticks(range(0, 10))

for x, y in zip(cores, speedup):
    ax2.annotate(f'{y:.2f}x', (x, y), textcoords="offset points", xytext=(0, -18), ha='center', fontsize=9, fontweight='bold', color='#d62728')

ax2.legend(loc='upper left', frameon=True)
ax2.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('speedup_vs_cores.png', dpi=300)
plt.show()
```
